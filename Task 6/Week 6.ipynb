```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("/content/drive/MyDrive/Colab Notebooks/healthcare_dataset (1).csv")
df.head()
```

```python
billing = df.groupby(["Medical Condition","Insurance Provider"])["Billing Amount"].sum().unstack()
billing
```

```python
billing.plot(
    kind = "bar",
    stacked = True,
    figsize = (10,6)
)
plt.title("Billing Amount by Medical Condition and Insurance Provider")
plt.xlabel("Medical Condition")
plt.ylabel("Billing Amount")
plt.show()
```

```python
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker

ax = billing.plot(kind="bar", stacked=True, figsize=(10, 6))

plt.title("Billing Amount by Medical Condition and Insurance Provider")
plt.xlabel("Medical Condition")
plt.ylabel("Billing Amount ($)")

plt.xticks(rotation=0)

plt.legend(title="Insurance Provider", bbox_to_anchor=(1.05, 1), loc="upper left")

ax.yaxis.set_major_formatter(ticker.FuncFormatter(lambda x, pos: f"${x*1e-6:.1f}M"))

plt.tight_layout()
plt.show()
```
