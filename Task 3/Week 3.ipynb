```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as pyplot
```

```python
df = pd.read_csv("/content/drive/MyDrive/Colab Notebooks/AAPL.csv")
df
```

```python
print(df.isnull().sum())
```

```python
avg_open = df['Open'].mean()
avg_close = df['Close'].mean()

print(f"Average Open Price:  ${avg_open:.2f}")
print(f"Average Close Price: ${avg_close:.2f}")
```

```python
df = df.sort_values(by='Date')
df
```

```python
df['Daily Delta'] = df['Close'] - df['Open']
```

```python
df['Daily Return'] = (
    (df['Close'] - df['Open'])/ df['Open']
    )*100
```

```python
print (df [["Date","Open","Close","Daily Delta","Daily Return"]
           ].head(10))
```

```python
average_volume = df["Volume"].mean()
maximum_volume = df["Volume"].max()
minimum_volume = df["Volume"].min()

print(f"Average Volume: {average_volume:.2f}")
print(f"Maximum Volume: {maximum_volume}")
print(f"Minimum Volume: {minimum_volume}")
```
